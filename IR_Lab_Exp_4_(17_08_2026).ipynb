```python
# 4. Classification of a set of Text Documents into known classes
# (You may use any of the Classification algorithms like Naive Bayes, Max Entropy, Rochio’s, Support Vector Machine).
# Standard Datasets will have to be used to show the results.

 # Classification of Text Documents using Naive Bayes

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Select categories
categories = [
    'rec.sport.baseball',
    'sci.space',
    'comp.graphics',
    'talk.politics.misc'
]

# Load training dataset
train_data = fetch_20newsgroups(
    subset='train',
    categories=categories,
    shuffle=True,
    random_state=42
)

# Load testing dataset
test_data = fetch_20newsgroups(
    subset='test',
    categories=categories,
    shuffle=True,
    random_state=42
)

# Convert text documents into TF-IDF features
vectorizer = TfidfVectorizer(
    stop_words='english',
    max_df=0.7
)

X_train = vectorizer.fit_transform(train_data.data)
X_test = vectorizer.transform(test_data.data)

# Create Naive Bayes classifier
classifier = MultinomialNB()

# Train the classifier
classifier.fit(X_train, train_data.target)

# Predict test documents
y_pred = classifier.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(test_data.target, y_pred)

print("Text Document Classification using Naive Bayes")
print("-----------------------------------------------")
print("Number of training documents:", len(train_data.data))
print("Number of testing documents:", len(test_data.data))
print("Number of classes:", len(categories))

print("\nClasses:")
for i, category in enumerate(categories):
    print(i, ":", category)

print("\nAccuracy:", accuracy)

# Classification report
print("\nClassification Report:")
print(classification_report(
    test_data.target,
    y_pred,
    target_names=test_data.target_names
))

# Test a new document
new_document = [
    "The spacecraft was launched into orbit to explore the planets and stars."
]

new_features = vectorizer.transform(new_document)
prediction = classifier.predict(new_features)

print("New Document:")
print(new_document[0])
print("\nPredicted Class:", test_data.target_names[prediction[0]])
```
