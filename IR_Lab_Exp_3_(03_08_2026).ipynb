```python
# 3. Construction of an Inverted Index for a given document collection comprising of at least 50 documents with a total vocabulary size of at least 1000 words.

# -----------------------------------------
# Construction of Inverted Index
# -----------------------------------------

# Step 1 : Upload ZIP file
from google.colab import files
uploaded = files.upload()

# Step 2 : Extract ZIP File
import zipfile
import os

zip_file = "documents_1000_vocab_real.zip"
extract_folder = "documents_1000_vocab_real"

os.makedirs(extract_folder, exist_ok=True)

with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall(extract_folder)

print("ZIP extracted successfully!")

# Step 3 : Display uploaded files
files_list = sorted(os.listdir(extract_folder))

print("\nNumber of Documents :", len(files_list))

print("\nFirst 10 Documents")

for file in files_list[:10]:
    print(file)

# Step 4 : Construct Inverted Index

import re
from collections import defaultdict

folder = "documents_1000_vocab_real"

stop_words = {
    "the","is","a","an","and","or","of","to","in","on","for",
    "with","by","as","at","from","that","this","it","are",
    "was","be","has","have"
}

inverted_index = defaultdict(set)

for filename in os.listdir(folder):

    if filename.endswith(".txt"):

        path = os.path.join(folder, filename)

        with open(path, "r", encoding="utf-8") as file:

            text = file.read().lower()

            # Remove punctuation
            text = re.sub(r'[^a-z0-9\s]', ' ', text)

            # Tokenization
            words = text.split()

            # Remove stop words and build index
            for word in words:

                if word not in stop_words:
                    inverted_index[word].add(filename)

# Step 5 : Display Inverted Index

print("\nINVERTED INDEX\n")

for term in sorted(inverted_index):
    print(f"{term} --> {sorted(inverted_index[term])}")

print("\nVocabulary Size :", len(inverted_index))
```
